```python
keyword = ['anjing', 'babi', 'monyet', 'kontol', 'ngentot', 'bangsat', 'jelek', 'goblok', 'tolol', 'cupu', 'bego', 'otak udang', 'dongo', 'mati aja', 'bajingan', 'ngaceng', 'lgbt']
```

```python
# Grouping keywords which have the same meaning
def grouping_keyword(text):
    for word in keyword:
        if word in text:
            return word
    return text

data['keyword_group'] = data['keyword'].apply(grouping_keyword)

data['keyword_group'].value_counts()
```

```python
# Count how many keywords are there
data['keyword_group'].nunique()
```

```python
# Filtering Indonesian tweets
data = data[data['lang'] == 'in']
data
```

```python
# Cleaning Text Function

def clean_text(text):
    text = re.sub(r'@\w+', '', text)
    text = re.sub(r"http\S+|www\S+", "", text)
    text = re.sub(r"#\w+", "", text)
    text = re.sub(r"\brt\b", "", text)
    text = re.sub(r"\d+", "", text)
    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r'[\n\r\t]+', ' ', text)
    text = re.sub(r"(\w+)\^2", r"\1 \1", text)
    text = re.sub(r'\s+', ' ', text).strip()
    text = re.sub(r'[\u1160\u3164]', '', text)
    text = re.sub(r'USER', '', text)

    return text.strip()
```

```python
# Remove Emoji Function
def remove_emoji(text):
    return emoji.replace_emoji(text, replace='') 
```

```python
# Preserve intentional uppercase words and normalize other capitalization

def smart_case(text):
    words = text.split()
    result = []
    for word in words:
        if word == 'USER' or word == '[URL]':
            result.append(word)
        elif len(word) >= 2 and word.isupper():
            result.append(word)
        else:
            result.append(word.lower())
    return ' '.join(result)
```

```python
def preprocess(text):

    text = clean_text(text)
    text = remove_emoji(text)
    text = smart_case(text)

    return text
```

```python
# Apply Cleaning Text
data['cleaned_text'] = data['text'].apply(preprocess)
```

```python
# Remove Strange Symbols
data['cleaned_text'] = data['cleaned_text'].replace(r'^\s*$', np.nan, regex=True)
```

```python
import json

with open('cleaned.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

with open('input_vertex.jsonl', 'w', encoding='utf-8') as f:
    for item in data:
        tweet = item['cleaned_text']
        structure = {
            "contents": [
                {
                    "role": "user",
                    "parts": [{"text": tweet}]
                }
            ]
        }
        f.write(json.dumps(structure, ensure_ascii=False) + '\n')

print("Selesai! File input_vertex.jsonl sudah siap.")
```

```python
# Insert prompt into the JSONL data

file_input_lama = "input_vertex.jsonl"  
file_output_vertex = "vertex.jsonl"

PROMPT_TEMPLATE = """Kamu adalah seorang ahli analisis teks dan moderator media sosial Indonesia. Tugasmu adalah mengklasifikasikan tweet ke dalam salah satu dari 5 kategori berdasarkan definisi berikut.

**PENTING:** Sebuah tweet HANYA bisa dikategorikan sebagai salah satu tipe Cyberbullying jika terdapat TARGET yang jelas di dalam teks seperti penyebutan USER maupun individu atau kelompok yang diserang/dituju. Jika tidak ada target spesifik, masukkan ke "Non-Cyberbullying".

KATEGORI DAN DEFINISI:
1. "Hate Speech" -> Ujaran secara publik yang mengekspresikan kebencian atau kekerasan terhadap seseorang maupun suatu kelompok berdasarkan ras, agama, jenis kelamin, atau orientasi seksual.
2. "Sexual Harassment" -> Teks atau pesan yang bersifat melecehkan secara seksual, atau merujuk pada aktivitas/foto seksual yang menyudutkan target.
3. "Kata Kasar dan Vulgar" -> Memberi pesan dengan kata-kata kasar, kotor, makian, atau bahasa vulgar yang menyerang dan dapat menyinggung orang lain.
4. "Body Shaming" -> Merundung penampilan fisik seseorang. Merupakan bentuk kekerasan verbal-emosional yang tujuannya beragam, mulai dari bahan candaan, memperkeruh suasana, hingga menghina target.
5. "Non-Cyberbullying" -> Tweet yang berupa opini kritis, keluhan umum, ekspresi emosi/kata kasar tanpa target personal, atau teks netral/positif.

ATURAN OUTPUT:
Berikan output HANYA dalam format JSON mentah dengan key "label" dan "alasan" singkat yang mendasari keputusanmu. Jangan berikan teks pengantar atau penutup apa pun selain JSON tersebut.

Tweet yang harus dinilai:
"{tweet_text}" """

# 3. Proses konversi data
print("Memulai konversi struktur JSONL...")
counter = 0

with open(file_input_lama, 'r', encoding='utf-8') as infile, \
     open(file_output_vertex, 'w', encoding='utf-8') as outfile:
    
    for line in infile:
        if not line.strip():
            continue
            
        data_lama = json.loads(line)
        tweet_asli = data_lama["contents"][0]["parts"][0]["text"]
        
        prompt_lengkap = PROMPT_TEMPLATE.format(tweet_text=tweet_asli)
        
        # Susun ulang untuk Vertex AI
        struktur_baru = {
            "request": {
                "contents": [
                    {
                        "role": "user",
                        "parts": [
                            {
                                "text": prompt_lengkap
                            }
                        ]
                    }
                ]
            }
        }
        
        outfile.write(json.dumps(struktur_baru, ensure_ascii=False) + '\n')
        counter += 1

print(f"Selesai! Berhasil memproses {counter} data tweet ke file baru: {file_output_vertex}")
```
